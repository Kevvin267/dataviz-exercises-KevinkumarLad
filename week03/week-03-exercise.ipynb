```python
import pandas as pd
import plotly.express as px
import plotly.graph_objects as go

# Dataset: CO2 Emissions by Country 2000-2022
# Source: Our World in Data (https://ourworldindata.org/co2-emissions)
df = pd.read_csv('/kaggle/input/datasets/ladkevinkumar/co2-emission-kevinkumar/co2_emissions.csv')
print(f"Loaded: {len(df)} rows | Countries: {df['Country'].nunique()} | Years: {df['Year'].min()}-{df['Year'].max()}")
print(df.head())
```

```python
# Explore before building

print("Countries:", df['Country'].unique())
print("\nCO2 range:", df['CO2_Mt'].min(), "to", df['CO2_Mt'].max(), "Mt")
print("\nRegional averages (2022):")
print(df[df['Year']==2022].groupby('Region')['CO2_Mt'].mean().sort_values(ascending=False).round(1))
```

```python
# Let's apply the principles

# Choose a country to highlight
highlight = 'United Kingdom'

# Build colour map: highlight in red, everything else grey
color_map = {c: 'crimson' if c == highlight else 'lightgrey' for c in df['Country'].unique()}


fig = px.line(df, x='Year', y='CO2_Mt', color='Country',
              color_discrete_map=color_map,
              labels={'CO2_Mt': 'CO2 Emissions (Mt)', 'Year': ''},
              title="United Kingdom's C02 emission is the lowest in 2022",
              template='plotly_white',
              height=400
              )
fig.update_traces(
    line=dict(width=1),   # default: thin for context countries
    showlegend=False     # direct label replaces legend (Gestalt proximity)
)

# Override highlighted country: thicker line to draw attention
fig.update_traces(
    line=dict(width=5),
    selector=dict(name=highlight) # each line is named after a country in the traces of the figure
)


# Direct label at end of highlighted line
highlight_label = df.loc[(df['Country'] == highlight) & (df['Year'] == df['Year'].max())]

others_label = df.loc[(df['Country'] == 'India') & (df['Year'] == df['Year'].max())]

# add highlight label
fig.add_annotation(
    x=highlight_label['Year'].values[0], y=highlight_label['CO2_Mt'].values[0],
    text=highlight, showarrow=False,
    xanchor='left', xshift=10,
    font=dict(color='crimson', size=12, weight='bold', family='Arial')
)

# add others label (optional)
fig.add_annotation(
    x=others_label['Year'].values[0], y=others_label['CO2_Mt'].values[0],
    text='Other Countries', showarrow=False,
    xanchor='left', xshift=10,
    font=dict(color='lightgrey', size=12, weight='bold', family='Arial')
)

fig.update_layout(
    yaxis=dict(gridcolor='#EEEEEE'),
    xaxis=dict(showgrid=False)
)



fig.show()
```

```python

import pandas as pd
import plotly.graph_objects as go

# Load dataset
df = pd.read_csv('/kaggle/input/datasets/ladkevinkumar/co2-emission-kevinkumar/co2_emissions.csv')

# Select Asian countries
asia_df = df[df["Region"] == "Asia"]

# Country to highlight
highlight_country = "China"

# Create figure
fig = go.Figure()

# Add grey lines for other Asian countries
for country in asia_df["Country"].unique():
    country_data = asia_df[asia_df["Country"] == country].sort_values("Year")

    if country == highlight_country:
        continue

    fig.add_trace(go.Scatter(
        x=country_data["Year"],
        y=country_data["CO2_Mt"],
        mode="lines",
        name=country,
        line=dict(color="lightgrey", width=2),
        showlegend=True
    ))

china_data = asia_df[asia_df["Country"] == highlight_country].sort_values("Year")

fig.add_trace(go.Scatter(
    x=china_data["Year"],
    y=china_data["CO2_Mt"],
    mode="lines",
    name="India (Highlighted)",
    line=dict(color="red", width=4)
))

# Chart layout
fig.update_layout(
    title="China's CO2 emission increase 3 times in 2022 according to 2000",
    xaxis_title="Year",
    yaxis_title="CO2 Emissions (Million Tonnes)",
    template="plotly_white",
    hovermode="x unified",
    width=1100,
    height=600
)

# Y-axis scale
fig.update_yaxes(
    rangemode="tozero",
    dtick=1000
)

fig.update_layout(
    yaxis=dict(gridcolor='#EEEEEE'),
    xaxis=dict(showgrid=False)
)



fig.show()
```
