```python
region_avg = (df.groupby('Region', as_index=False)['Happiness_Score']
              .mean().sort_values('Happiness_Score', ascending=True))
display(region_avg)
top_region = region_avg.iloc[-1]['Region']
top_score = region_avg.iloc[-1]['Happiness_Score']
fig1 = px.bar(region_avg, x='Happiness_Score', y='Region', orientation='h',
              text='Happiness_Score', color_discrete_sequence=['#3167A8'],
              title=f'{top_region} leads regional happiness averages ({top_score:.2f})')
fig1.update_traces(texttemplate='%{text:.2f}', textposition='outside', cliponaxis=False)
fig1.update_layout(template='plotly_white', height=570, title_x=0.03,
                   xaxis_title='Average happiness score', yaxis_title='',
                   showlegend=False, margin=dict(l=80, r=85, t=85, b=65))
fig1.update_xaxes(range=[0, 9], rangemode='tozero', showgrid=True, gridcolor='#E8E8E8')
fig1.show()
```

```python
top8 = df.nlargest(8, 'Happiness_Score').copy()
top8['Group'] = 'Top 8'
bottom8 = df.nsmallest(8, 'Happiness_Score').copy()
bottom8['Group'] = 'Bottom 8'
combined = pd.concat([bottom8, top8], ignore_index=True).sort_values('Happiness_Score')
global_avg = df['Happiness_Score'].mean()
gap = top8.iloc[0]['Happiness_Score'] - bottom8.iloc[-1]['Happiness_Score']
print('Global average:', round(global_avg, 2), '| Highest-to-lowest gap:', round(gap, 2))
fig2 = px.bar(combined, x='Happiness_Score', y='Country', orientation='h',
              color='Group', color_discrete_map={'Top 8':'#277A68', 'Bottom 8':'#D46A52'},
              text='Happiness_Score',
              title=f'Finland and Afghanistan are {gap:.2f} points apart in happiness score',
              category_orders={'Country':combined['Country'].tolist()})
fig2.update_traces(texttemplate='%{text:.2f}', textposition='outside', cliponaxis=False)
fig2.add_vline(x=global_avg, line_dash='dash', line_color='#565656', line_width=1.5,
               annotation_text=f'Global mean: {global_avg:.2f}', annotation_position='top')
fig2.add_hline(y=7.5, line_dash='dot', line_color='#9BA6B2', line_width=1)
fig2.update_layout(template='plotly_white', height=750, title_x=0.02,
                   xaxis_title='Happiness score', yaxis_title='',
                   legend_title_text='', margin=dict(l=75,r=100,t=100,b=60))
fig2.update_xaxes(range=[0,9], rangemode='tozero', gridcolor='#E8E8E8')
fig2.show()
```

```python

import plotly.express as px

regions_to_plot = [
    'Western Europe',
    'Latin America and Caribbean',
    'East Asia',
    'Sub-Saharan Africa',
    'South Asia'
]

regional = (
    df[df['Region'].isin(regions_to_plot)]
    .groupby('Region')[['GDP', 'Freedom']]
    .mean()
)

normalized = regional.div(df[['GDP', 'Freedom']].mean())

long = normalized.reset_index().melt(
    id_vars='Region',
    var_name='Factor',
    value_name='Relative to Global Mean'
)

fig3 = px.bar(
    long,
    x='Relative to Global Mean',
    y='Region',
    color='Factor',
    orientation='h',
    barmode='group',
    color_discrete_map={
        'GDP': '#315D9C',
        'Freedom': '#E69A3A'
    },
    title='GDP and Freedom Contributions Vary by Region',
    category_orders={'Region': regions_to_plot}
)

fig3.update_layout(
    template='plotly_white',
    height=560,
    xaxis_title='Ratio to Global Mean (1 = Global Mean)',
    yaxis_title='',
    legend_title='Factor',
    bargap=0.25
)

fig3.update_xaxes(rangemode='tozero')

fig3.add_vline(
    x=1,
    line_dash='dash',
    line_color='#666666'
)

fig3.show()
```
